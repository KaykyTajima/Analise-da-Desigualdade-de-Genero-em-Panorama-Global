```python
df_sigi = lista_de_dataframes_extras[0] #Social Institutions and Gender Index

df_ggi = lista_de_dataframes_extras[1] #Global Gender Index

#Dados UNDP - tem as mesmas colunas, indicadores diferentes
df_gii = lista_de_dataframes_extras[2] #Gender Inequalities Index (UNDP)
df_gdi = lista_de_dataframes_extras[3] #Gender Development Index (UNDP)

df_undp = pd.concat([df_gii, df_gdi])

ano = 2022 #Vamos testar apenas para o ano de 2023
```

```python
# df_undp - GII e GDI
# GII - Gender Inequalities Index
# GDI - Gender Development Index
# Dados undp precisam ser pivotados, estão em formato longo

df_filtrado = df_undp[
        (df_undp['year'] == ano) &
        (df_undp['indicatorCode'].isin(['gii', 'gdi']))
    ]

df_undp_pivot = df_filtrado.pivot_table(
        index='countryIsoCode',
        columns='indicatorCode',
        values='value'
    ).reset_index()

colunas_renomear = {
        'countryIsoCode': 'ISO_Code',
        'gii': 'GII_Score',
        'gdi': 'GDI_Score'
    }

df_undp_final = df_undp_pivot.rename(columns=colunas_renomear).set_index('ISO_Code')
df_undp_final
```

```python
# GGI - Gender Gap Index
#Overall Index: gggi_ggi
#Economic: gggi_pos (Participation and Opportunity Subindex)
#Education: gggi_eas (Educational Attainment Subindex)
#Health: gggi_hss (Health and Survival Subindex)
#Political: gggi_pes (Political Empowerment Subindex)
```

```python
df_ggi_filtrado = df_ggi[df_ggi['year'] == ano]

colunas_renomear = {
        'ccodealp': 'ISO_Code',
        'gggi_ggi': 'GGGI_Score',  # Score Geral
        'gggi_pos': 'GGGI_Eco',    # Economic (Particip. & Oppor.)
        'gggi_eas': 'GGGI_Edu',    # Education (Attainment)
        'gggi_hss': 'GGGI_Health', # Health
        'gggi_pes': 'GGGI_Pol'     # Political
    }

colunas_desejadas = list(colunas_renomear.keys())

df_ggi_final = (df_ggi_filtrado
          .filter(items=colunas_desejadas)
          .rename(columns=colunas_renomear)
          .set_index('ISO_Code')
    )
df_ggi_filtrado
```

```python
df_sigi
```

```python
#SIGI
nomes_indicadores_sigi = [
        'SIGI score',
        'Discrimination in the family',
        'Restricted physical integrity',
        'Access to productive and financial assets',
        'Restricted civil liberties'
    ]

df_filtrado = df_sigi[
    (df_sigi['TIME_PERIOD'] == 2023) &
        (df_sigi['Measure'].isin(nomes_indicadores_sigi))
    ]

df_pivot = df_filtrado.pivot_table(
        index='REF_AREA',
        columns='Measure',
        values='OBS_VALUE'
    ).reset_index()

colunas_renomear = {
        'REF_AREA': 'ISO_Code',
        'SIGI score': 'SIGI_Score',
        'Discrimination in the family': 'SIGI_Family',
        'Restricted physical integrity': 'SIGI_Integrity',
        'Access to productive and financial assets': 'SIGI_Resources',
        'Restricted civil liberties': 'SIGI_Civil'
    }

df_sigi_final = df_pivot.rename(columns=colunas_renomear).set_index('ISO_Code')
df_sigi_final
```

```python
#Unificar em um dataframe
df_list = [df_undp_final, df_ggi_final, df_sigi_final]
df_merged = pd.concat(df_list, axis=1, join='inner')
```

```python
df_merged
```

```python
# Transformação dos Dados
df_analysis = df_merged.copy()
df_analysis['GII_Equidade'] = 1 - df_analysis['GII_Score']
df_analysis['SIGI_Equidade'] = 100 - df_analysis['SIGI_Score']
```

```python
# --- Análise 1: Correlação dos Índices Principais ---
print("\nGerando Heatmap 1: Correlação dos Índices Principais...")
colunas_principais = ['GII_Equidade', 'GDI_Score', 'GGGI_Score', 'SIGI_Equidade']
colunas_principais_validas = [col for col in colunas_principais if col in df_analysis.columns]

corr_matrix_main = df_analysis[colunas_principais_validas].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(
     corr_matrix_main,
     annot=True, cmap='vlag', fmt='.2f', linewidths=0.5, vmin=-1, vmax=1
)
plt.title('Heatmap 1: Correlação entre os Principais Índices de Gênero\n(Valores altos = Alta Equidade)', fontsize=16)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# --- Análise 2: Correlação dos Sub-Índices ---

# Renomear colunas de sub-índice
df_analysis['SIGI_Family_Eq'] = 100 - df_analysis['SIGI_Family']
df_analysis['SIGI_Integrity_Eq'] = 100 - df_analysis['SIGI_Integrity']
df_analysis['SIGI_Resources_Eq'] = 100 - df_analysis['SIGI_Resources']
df_analysis['SIGI_Civil_Eq'] = 100 - df_analysis['SIGI_Civil']

# Lista de colunas final (incluindo GGGI_Edu)
colunas_subindices = [
        'GII_Equidade',
        'GDI_Score',
        'GGGI_Eco',
        'GGGI_Edu',
        'GGGI_Health',
        'GGGI_Pol',
        'SIGI_Family_Eq',
        'SIGI_Integrity_Eq',
        'SIGI_Resources_Eq',
        'SIGI_Civil_Eq'
    ]

colunas_subindices_validas = [col for col in colunas_subindices if col in df_analysis.columns]
print(f"\nColunas usadas no Heatmap 2: {colunas_subindices_validas}")

df_sub = df_analysis[colunas_subindices_validas]

corr_matrix_sub = df_sub.corr()

plt.figure(figsize=(14, 12))
sns.heatmap(
        corr_matrix_sub,
        annot=True, cmap='vlag', fmt='.2f', linewidths=0.5, vmin=-1, vmax=1
)
plt.title('Heatmap 2: Correlação com Sub-Índices (WEF & OECD)\n(Valores altos = Alta Equidade)', fontsize=16)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
def run_model_and_plot_importance(X, y, target_name):
    """Treina um RandomForest, avalia e plota a importância das features."""
    print(f"\n--- Treinando modelo para: {target_name} ---")
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
    model = RandomForestRegressor(n_estimators=100, random_state=42, oob_score=True)
    model.fit(X_train, y_train)
    preds = model.predict(X_test)
    r2 = r2_score(y_test, preds)
    oob = model.oob_score_
    print(f"Score R² (Teste): {r2:.3f}")
    print(f"Score OOB (Out-of-Bag): {oob:.3f}")

    importances = model.feature_importances_
    feature_names = X.columns
    df_importance = pd.DataFrame({'Feature': feature_names, 'Importance': importances})
    df_importance = df_importance.sort_values(by='Importance', ascending=False)

    plt.figure(figsize=(12, 8))
    sns.barplot(x='Importance', y='Feature', data=df_importance, palette='viridis')
    plt.title(f'Importância das Features para prever: {target_name}', fontsize=16)
    plt.xlabel('Importância (Gini Importance)')
    plt.ylabel('Preditores (Features)')
    plt.tight_layout()
    plt.show()

    return r2
```

```python
X = df_model_clean[features_list]
y_eco = df_model_clean[target_eco]
run_model_and_plot_importance(X, y_eco, target_name="Pilar Econômico (GGGI_Eco)")

X_pol = X.copy()
y_pol = df_model_clean[target_pol]
run_model_and_plot_importance(X_pol, y_pol, target_name="Pilar Político (GGGI_Pol)")
```

```python
import pandas as pd
import numpy as np # Necessário para NaN e Inf
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.impute import SimpleImputer # Para tratar NaNs
from sklearn.preprocessing import StandardScaler # Para escalar os dados
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
df_list = [df_undp_final, df_ggi_final, df_sigi_final, df_renomeado]
df_master = pd.concat(df_list, axis=1, join='inner')

print(f"\n--- Merge completo ---")
if df_master.empty: print("!!! ERRO: Merge vazio.");
print(f"Total de {len(df_master)} países com dados nos datasets base (antes da limpeza).")
```

```python
variaveis_cluster = [
        'WDI_GDP_Capita', # Riqueza
        'GDI_Score',      # Paridade Desenvolvimento Humano Básico
        'SIGI_Score',     # Normas Institucionais (Alto = Ruim)
        'GGGI_Eco',       # Gap Econômico (Alto = Bom)
        'GGGI_Pol'        # Gap Político (Alto = Bom)
    ]
# Filtra apenas as que existem
variaveis_validas = [v for v in variaveis_cluster if v in df_master.columns]
print(f"\nVariáveis usadas para clusterização: {variaveis_validas}")

df_cluster_raw = df_master[variaveis_validas].copy()
```

```python
print(f"\nNúmero de países ANTES da imputação: {len(df_cluster_raw)}")
print("Valores ausentes por coluna ANTES da imputação:")
print(df_cluster_raw.isnull().sum())

# Usaremos a Mediana para imputar, pois é menos sensível a outliers que a Média
imputer = SimpleImputer(strategy='median')
# O imputer retorna um array numpy, precisamos converter de volta para DataFrame
df_cluster_imputed_np = imputer.fit_transform(df_cluster_raw)
df_cluster_imputed = pd.DataFrame(df_cluster_imputed_np,
                                     columns=df_cluster_raw.columns,
                                     index=df_cluster_raw.index)

print(f"\nNúmero de países DEPOIS da imputação: {len(df_cluster_imputed)}")
print("Valores ausentes por coluna DEPOIS da imputação:")
print(df_cluster_imputed.isnull().sum()) # Deve ser tudo zero
```

```python
scaler = StandardScaler()
df_cluster_scaled_np = scaler.fit_transform(df_cluster_imputed)
df_cluster_scaled = pd.DataFrame(df_cluster_scaled_np,
                                     columns=df_cluster_imputed.columns,
                                     index=df_cluster_imputed.index)
print("\nDados normalizados (primeiras 5 linhas):")
print(df_cluster_scaled.head())
```

```python
wcss = [] # Within-cluster sum of squares (para Elbow Method)
silhouette_scores = [] # Para Silhouette Score
k_valid_for_silhouette = [] # Guarda os 'k' para os quais a silhueta foi calculada

k_range = range(2, 11)

print(f"Número de países (samples) entrando no K-Means: {len(df_cluster_scaled)}") # Diagnóstico

for k in k_range:
  kmeans = KMeans(n_clusters=k, init='k-means++', random_state=42, n_init=10)
  # kmeans.fit(df_cluster_scaled) # Mudança: fit_predict abaixo faz o fit
  labels = kmeans.fit_predict(df_cluster_scaled) # Usar fit_predict para pegar os labels

  # Calcula WCSS (sempre possível)
  wcss.append(kmeans.inertia_)

  # Verifica quantos clusters únicos foram REALMENTE encontrados
  n_clusters_found = len(np.unique(labels))

  # Calcula Silhouette Score SOMENTE se encontrar 2 ou mais clusters
  if n_clusters_found >= 2:
     score = silhouette_score(df_cluster_scaled, labels)
     silhouette_scores.append(score)
     k_valid_for_silhouette.append(k) # Guarda o k válido
     print(f"  k={k}: WCSS={kmeans.inertia_:.2f}, Silhouette={score:.3f} ({n_clusters_found} clusters encontrados)")
  else:
     # Se encontrou menos de 2 clusters, não calcula silhueta
     # Adiciona um valor placeholder (NaN) para manter o tamanho da lista (opcional, mas ajuda no plot)
     # silhouette_scores.append(np.nan) # Removido para simplificar o plot
     print(f"  k={k}: WCSS={kmeans.inertia_:.2f}, Silhouette=N/A ({n_clusters_found} cluster encontrado, não calculável)")


# Plotar Elbow Method (Funciona mesmo se a silhueta falhar)
plt.figure(figsize=(10, 5))
plt.plot(k_range, wcss, marker='o', linestyle='--')
plt.xlabel('Número de Clusters (k)')
plt.ylabel('WCSS (Inertia)')
plt.title('Método do Cotovelo (Elbow Method)')
plt.xticks(k_range)
plt.grid(True)
plt.show()

# Plotar Silhouette Scores (Apenas para os k válidos)
if silhouette_scores: # Verifica se a lista não está vazia
   plt.figure(figsize=(10, 5))
   plt.plot(k_valid_for_silhouette, silhouette_scores, marker='o', linestyle='--') # Usa k_valid_for_silhouette no eixo X
   plt.xlabel('Número de Clusters (k)')
   plt.ylabel('Silhouette Score Médio')
   plt.title('Coeficiente de Silhueta para K-Means')
   plt.xticks(k_valid_for_silhouette) # Mostra apenas os k válidos
   plt.grid(True)
   plt.show()
else:
    print("\nNão foi possível calcular o Silhouette Score para nenhum valor de k testado.")
```

```python
#Definição do k
# Pelo grafico do cotovelo - 4 ou 5
# Pelo grafico da silhueta - 2
# Separar em apenas 2 pode ser muito simplista para a variabilidade dos dados que temos
# Ir no 4 ou 5
```

```python
k_escolhido = 5
print(f"\nExecutando K-Means com k={k_escolhido}...")

kmeans_final = KMeans(n_clusters=k_escolhido, init='k-means++', random_state=42, n_init=10)
clusters = kmeans_final.fit_predict(df_cluster_scaled)

# Adicionar os labels dos clusters ao DataFrame (não escalonado, para interpretação)
df_final_clusters = df_cluster_imputed.copy()
df_final_clusters['Cluster'] = clusters
df_final_clusters['Cluster'] = df_final_clusters['Cluster'].astype('category') # Melhora a ordenação
```
